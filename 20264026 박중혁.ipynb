```python
# <예제 10-1>
# pandas, matplotlib 예시
import pandas as pd
import matplotlib.pyplot as plt

# 예제 데이터프레임 생성
data = {
    '이름': ['홍길동', '이몽룡', '성춘향'],
    '국어': [95, 80, 90],
    '수학': [75, 85, 100]
}

df = pd.DataFrame(data)

# '국어' 점수의 평균 계산
국어_평균 = df['국어'].mean()
print(f"국어 점수의 평균: {국어_평균}")

# 데이터프레임을 바탕으로 막대 그래프 그리기
df.plot(kind='bar', x='이름', y=['국어', '수학'], title='학생별 점수')
plt.ylabel('점수')
plt.xlabel('학생 이름')
plt.legend(title='과목')
plt.show()
```

```python
# 예제 <10-6>
# PyTorch를 사용한 간단한 모델 구축

import torch
import torch.nn as nn
import torch.nn.functional as F

# 간단한 신경망 정의
class SimpleModel(nn.Module):
    def __init__(self):
        super(SimpleModel, self).__init__()
        self.fc1 = nn.Linear(32, 64)
        self.fc2 = nn.Linear(64, 10)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x

# 모델 인스턴스 생성
model_pt = SimpleModel()

# 모델 구조 출력
print(model_pt)
```
